# Reciprocal space: Ewald sphere and reciprocal lattice

Ewald construction in the **lab frame** for a set of diffractometer angles:

- the reciprocal lattice (blue points, size ~ $(|F|^2)^{1/3}$, systematic absences hidden) rotates with the sample angles (mu, eta, chi, phi); (000) is white,
- the Ewald sphere (yellow wireframe, radius $k = 2\pi/\lambda$) is fixed in the lab, centred at $-k_i$ from (000),
- $k_i$ (yellow) and $k_f$ (orange) start at the sphere centre, $Q = k_f - k_i$ (magenta) at (000); $k_f$ is set by the detector angles (delta, gamma),
- reflections within `tolerance` of the sphere are in diffraction condition: red halo, listed by `vis.in_condition()`,
- a\*, b\*, c\* red / green / blue (to scale), SwissFEL axes triad dark red / green / blue (x, y, z).

The default camera looks from the left (SwissFEL +x) onto the vertical scattering plane.
Set `BROWSER = "headless"` to render without a window and save images with `vis.snapshot(path)`.

```python
from diffraction import Crystal
from diffraction.visualization.widgets import angle_controls, reciprocal_view_widget

BROWSER = "notebook"   # or None (browser tab) / "headless" (no window, use vis.snapshot)
```

## Crystal: c-axis normal up, a\* along the beam (SwissFEL frame)

```python
crystal = Crystal.from_cif("Bi2Se3", "diffraction/data/Bi2Se3.cif")
crystal.add_orientation((0, 0, 1), (0, 1, 0), "normal up")
crystal.add_orientation((1, 0, 0), (0, 0, 1), "a* along beam")
crystal.calc_ub()
crystal.set_surface_normal((0, 0, 1))
crystal.set_constraints(mu=0, chi=0, phi=0)

energy = 7500
df = crystal.calc_angles((0, 0, 15), energy)
df[["mu", "eta", "chi", "phi", "gamma", "delta"]]
```

## Ewald construction for one solution

Options: `q_max` (largest |Q| drawn, default 2k = all reachable reflections), `min_intensity` (relative cut-off),
`max_points`, `tolerance` (excitation error in 1/Å, default 1 % of k), `limiting_sphere=True`.

```python
vis = crystal.show_reciprocal(df.iloc[0], energy, browser=BROWSER)
print(vis.summary())
```

Reflections in diffraction condition and the drawn reciprocal lattice as tables:

```python
vis.in_condition()[["h", "k", "l", "q", "rel_intensity", "excitation_error"]]
```

```python
vis.points.nlargest(10, "intensity")[["h", "k", "l", "q", "intensity", "rel_intensity"]]
```

Rotating the sample moves the lattice through the sphere:

```python
vis.set_angles(eta=20)
print(vis.summary())
```

## Interactive controls

Pick a `calc_angles` solution for any hkl, then move the angles with the sliders (updates on release).

```python
vis, controls = reciprocal_view_widget(crystal, energy, hkl=(1, 0, 10), vis=vis)
controls
```
